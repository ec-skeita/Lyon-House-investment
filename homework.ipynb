```python
def calculate_savings(annual_salary, savings_rate, semi_annual_raise, annual_return, months):

    current_savings = 0.0
    current_annual_salary = annual_salary
    monthly_return_rate = annual_return / 12
    
    for month in range(months):
        if month > 0 and month % 6 == 0:
            current_annual_salary *= (1 + semi_annual_raise)
        
        monthly_salary = current_annual_salary / 12
        monthly_contribution = monthly_salary * savings_rate
        investment_return = current_savings * monthly_return_rate
        
        current_savings += monthly_contribution + investment_return
    
    return current_savings

if __name__ == "__main__":
    find_savings_rate()
```
